```python
import ipywidgets as widgets
from IPython.display import HTML, display

display(HTML("""
<style>
.resposta-escura {
    background-color: #1f2937 !important;
    color: #f3f4f6 !important;
    border: 1px solid #374151;
    border-left: 4px solid #60a5fa;
    border-radius: 6px;
    padding: 10px;
}
.resposta-escura .widget-label,
.resposta-escura label {
    color: #f3f4f6 !important;
}
.resposta-escura textarea {
    background-color: #111827 !important;
    color: #f9fafb !important;
    border: 1px solid #4b5563 !important;
    border-radius: 5px !important;
    caret-color: #f9fafb !important;
}
.resposta-escura textarea::placeholder {
    color: #9ca3af !important;
}
.resposta-escura input[type='checkbox'] {
    accent-color: #60a5fa;
}
</style>
"""))

def campo_resposta(numero):
    erro = widgets.Checkbox(
        value=False,
        description="Esta linha contém erro",
        indent=False,
    )
    justificativa = widgets.Textarea(
        value="",
        placeholder="Digite aqui a justificativa da sua escolha...",
        description="Justificativa:",
        style={"description_width": "initial"},
        layout=widgets.Layout(width="98%", height="82px"),
    )
    bloco = widgets.VBox(
        [erro, justificativa],
        layout=widgets.Layout(width="100%", margin="0 0 12px 0"),
    )
    bloco.add_class("resposta-escura")
    display(bloco)
    return erro, justificativa
```
