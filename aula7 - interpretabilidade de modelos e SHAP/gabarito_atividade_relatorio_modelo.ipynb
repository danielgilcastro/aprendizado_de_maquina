## Síntese conceitual

Os problemas se concentram em seis ideias:

1. objetivo, saída e uso pretendido não estão alinhados;
2. ausência de `NaN` não significa ausência de dados desconhecidos;
3. `duration` provoca vazamento temporal e não permite conclusão causal;
4. acurácia isolada, um único particionamento e seleção pelo teste não bastam;
5. importância de variável e limiar de decisão exigem validação;
6. retirar uma variável sensível não elimina proxies nem garante justiça ou generalização.
